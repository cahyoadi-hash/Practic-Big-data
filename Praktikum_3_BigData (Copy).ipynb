## EKSPLORASI HANDS ON



```python
!hdfs dfs -mkdir -p /user/mahasiswa/eksplor
!hdfs dfs -ls /user/mahasiswa
```

```python
# 1. Buat file lokal pertama
with open("file1.txt", "w") as f:
    f.write("ini file 1\n")

# 2. Upload ke folder eksplor di HDFS
!hdfs dfs -put file1.txt /user/mahasiswa/eksplor/

# 3. Buat file lokal kedua
with open("file2.txt", "w") as f:
    f.write("ini file 2\n")

# 4. Gabungkan isi file2 ke file1 yang ada di HDFS
!hdfs dfs -appendToFile file2.txt /user/mahasiswa/eksplor/file1.txt

# 5. Cek hasilnya (isinya akan gabung jadi dua baris)
!hdfs dfs -cat /user/mahasiswa/eksplor/file1.txt
```

```python
# 1. Salin file1.txt menjadi file_salinan.txt di dalam HDFS
!hdfs dfs -cp /user/mahasiswa/eksplor/file1.txt /user/mahasiswa/eksplor/file_salinan.txt

# 2. Ubah nama file_salinan.txt menjadi file_baru.txt
!hdfs dfs -mv /user/mahasiswa/eksplor/file_salinan.txt /user/mahasiswa/eksplor/file_baru.txt

# 3. Tampilkan isi folder untuk memastikan file_baru.txt sudah ada
!hdfs dfs -ls /user/mahasiswa/eksplor
```

```python
# Membuat file kosong 
!hdfs dfs -touchz /user/mahasiswa/eksplor/file_kosong.txt

# Verifikasi hasil akhir semua file 
!hdfs dfs -ls /user/mahasiswa/eksplor
```
