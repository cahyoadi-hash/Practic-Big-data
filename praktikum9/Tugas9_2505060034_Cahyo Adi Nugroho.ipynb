```python
#Soal A

from pymongo import MongoClient

# Membuat koneksi dan mendefinisikan database serta collection
client = MongoClient("mongodb://localhost:27017/")
db = client["kopi_nusantara"]
koleksi_menu = db["menu"]

#menghapus seluruh data agar data tidak duplikat saat cell ini di run berkali-kali
koleksi_menu.delete_many({})

#Menambahkan data
koleksi_menu.insert_many([
    # Kategori Kopi
    {"nama_menu": "Espresso Single", "harga": 15000, "kategori": "Kopi", "tingkat_kepahitan": 5, "rekomendasi_barista": True},
    {"nama_menu": "Americano Dingin", "harga": 18000, "kategori": "Kopi", "tingkat_kepahitan": 4, "stok": 50},
    {"nama_menu": "Cappuccino", "harga": 25000, "kategori": "Kopi", "tingkat_kepahitan": 3},
    {"nama_menu": "Kopi Susu Gula Aren", "harga": 22000, "kategori": "Kopi", "tingkat_kepahitan": 2, "stok": 120},
    {"nama_menu": "Caramel Macchiato", "harga": 28000, "kategori": "Kopi", "tingkat_kepahitan": 1, "rekomendasi_barista": True},
    {"nama_menu": "Latte", "harga": 24000, "kategori": "Kopi", "tingkat_kepahitan": 2},
    # Kategori Makanan
    {"nama_menu": "Ayam Geprek Kampus", "harga": 20000, "kategori": "Makanan", "pedas_level": 3, "stok": 30},
    {"nama_menu": "Mie Goreng Nyemek", "harga": 18000, "kategori": "Makanan", "pedas_level": 2},
    {"nama_menu": "Nasi Gila", "harga": 22000, "kategori": "Makanan", "pedas_level": 2, "rekomendasi_barista": True},
    {"nama_menu": "Spaghetti", "harga": 25000, "kategori": "Makanan", "pedas_level": 1},
    {"nama_menu": "Kentang Goreng", "harga": 20000, "kategori": "Makanan", "pedas_level": 0},
    {"nama_menu": "Roti Bakar Cokelat", "harga": 15000, "kategori": "Makanan", "pedas_level": 0, "stok": 40}
])

print("Jumlah dokumen di collection 'menu':", koleksi_menu.count_documents({}))
```

```python
#Soal B
#menu kategori "Kopi" dengan tingkat_kepahitan di atas 3
print("=== 1. Menu Kopi dengan tingkat kepahitan 3 ===")
kueri_1 = koleksi_menu.find({
    "kategori": "Kopi", 
    "tingkat_kepahitan": {"$gt": 3}
})
for menu in kueri_1:
    print(f"- {menu['nama_menu']} (Tingkat Kepahitan: {menu['tingkat_kepahitan']})")

print("\n")

#menu dengan harga di antara 15000 dan 30000
print("=== 2. Menu dengan harga antara 15.000 - 30.000 ===")
kueri_2 = koleksi_menu.find({
    "harga": {"$gte": 15000, "$lte": 30000}
})
for menu in kueri_2:
    print(f"- {menu['nama_menu']} (Harga: Rp{menu['harga']:,})")

print("\n")

#Menu yang tidak memiliki field pedas_level sama sekali
print("=== 3. Menu yang tidak memiliki field 'pedas_level' ===")
kueri_3 = koleksi_menu.find({
    "pedas_level": {"$exists": False}
})
for menu in kueri_3:
    print(f"- {menu['nama_menu']} (Kategori: {menu['kategori']})")
```

```python
#Soal C
#field status "Tersedia"
koleksi_menu.update_many(
    {}, 
    {"$set": {"status": "Tersedia"}}
)
#status menjadi "Habis" untuk 2 menu pilihan
koleksi_menu.update_one(
    {"nama_menu": "Espresso Single"}, 
    {"$set": {"status": "Habis"}}
)

koleksi_menu.update_one(
    {"nama_menu": "Nasi Gila"}, 
    {"$set": {"status": "Habis"}}
)
print("Status  'Espresso Single' dan 'Nasi Gila'  diubah menjadi Habis.\n")

print("=== Daftar Menu dan Status Ketersediaan ===")
for menu in koleksi_menu.find({}, {"_id": 0, "nama_menu": 1, "status": 1}):
    print(f"- {menu['nama_menu']}: {menu.get('status')}")
```

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import avg
import pandas as pd

#SparkSession
spark = SparkSession.builder \
    .appName("Tugas9") \
    .master("local[*]") \
    .getOrCreate()
spark.sparkContext.setLogLevel("ERROR")

#ekstrak seluruh dokumen dari collection menu
dokumen_menu = list(koleksi_menu.find({}))

#Mengubah data menjadi Pandas
df_pandas = pd.DataFrame(dokumen_menu)

#Menyesuaikan tipe data
df_pandas["_id"] = df_pandas["_id"].astype(str)
df_pandas = df_pandas.where(pd.notna(df_pandas), None)

#Mengubah Pandas menjadi Spark
df_spark_menu = spark.createDataFrame(df_pandas)

print("=== Skema Hasil Konversi ===")
df_spark_menu.printSchema()

print("=== Rata-rata Harga per Kategori ===")
df_spark_menu.groupBy("kategori").agg(
    avg("harga").alias("rata_rata_harga")
).show()

spark.stop()
```

apa tantangan yang anda temui saat mengonversi data MongoDB (dengan field yang tidak seragam antar dokumen) menjadi Spark DataFrame (yang mengharuskan skema tabel seragam)? Bagaimana anda mengatasinya?


**Jawaban =**

Dalam proses integrasi data dari MongoDB ke Apache Spark, tantangan utama yang saya hadapi bermula dari perbedaan arsitektur skema kedua sistem tersebut. MongoDB menggunakan arsitektur basis data NoSQL, yang memungkinkan setiap dokumen memiliki field atau atribut yang tidak seragam antar-barisnya misalnya, keberadaan field rekomendasi_barista atau stok yang hanya dimiliki oleh sebagian dokumen menu. Sebaliknya, Spark DataFrame menggunakan  pendekatan yang mensyaratkan skema tabel harus seragam. Ketika data yang tidak seragam ini diekstrak dari MongoDB dan diubah menjadi Pandas DataFrame sebagai perantara, Pandas secara otomatis akan merespons absennya suatu field dengan mengisinya menggunakan nilai NaN (Not a Number) yang berklasifikasi tipe data numerik (float). Hal ini memicu kendala berupa PySparkTypeError saat Pandas DataFrame tersebut dikonversi menjadi Spark DataFrame. Mesin PySpark akan menolak pemrosesan karena mendeteksi adanya percampuran tipe data yang bertentangan di dalam satu kolom yang sama, contohnya percampuran antara nilai True (boolean) dan NaN (float). Untuk mengatasi ini saya menimpa seluruh nilai NaN menjadi tipe None bawaan Python. Dengan begini, spark dapat membaca sel yang kosong tersebut dengan aman sebagai NULL.
